Описание задачи:
Создать базовый класс Inventory в C#, который будет представлять информацию о
наличии товаров на складе. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Inventory:
• Атрибуты: ID склада (WarehouseId), Название склада (WarehouseName),
Общий объем хранения (StorageCapacity).
• Методы:
o
o GetStorageStatus(): метод для получения статуса доступного
пространства на складе.
o AddItem(Item item): метод для добавления товара на склад.
o RemoveItem(Item item): метод для удаления товара со склада.
Требования к производным классам:
1. ПерсональныйСклад (PersonalInventory): Должен содержать
дополнительные атрибуты, такие как Владелец склада (OwnerName).
Метод GetStorageStatus() должен быть переопределен для отображения
информации о владельце склада вместе с статусом хранения.
2. ГрупповойСклад (GroupInventory): Должен содержать дополнительные
атрибуты, такие как Группа товаров (ProductGroup). Метод AddItem() должен
быть переопределен для добавления информации о группе товаров при
добавлении нового товара.
3. АвтоматизированныйСклад (AutomatedInventory) (если требуется третий
класс): Должен содержать дополнительные атрибуты, такие как
Автоматизация уровня (AutomationLevel). Метод RemoveItem() должен быть
переопределен для добавления информации о уровне автоматизации при
удалении товара.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;

class Item
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Manufacturer { get; set; }
    public int Weight { get; set; }

    public Item(string name, decimal price, string manufacturer, int weight)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        Weight = weight;
    }

    public void ShowInfo()
    {
        Console.WriteLine($"Товар: {Name}, цена: {Price} руб.");
    }

    public void ShowInfo(bool showManufacturer)
    {
        ShowInfo();

        if (showManufacturer)
        {
            Console.WriteLine($"Производитель: {Manufacturer}");
        }
    }

    public void ChangePrice(decimal newPrice)
    {
        Price = newPrice;
    }

    public decimal GetTotalPrice(int quantity)
    {
        return Price * quantity;
    }
}

class Inventory
{
    public int WarehouseId { get; set; }
    public string WarehouseName { get; set; }
    public int StorageCapacity { get; set; }
    protected int CurrentItems { get; set; }

    public string Address { get; set; }
    public string ManagerName { get; set; }
    public bool IsOpen { get; set; }

    public Inventory(int id, string name, int capacity)
    {
        WarehouseId = id;
        WarehouseName = name;
        StorageCapacity = capacity;
        CurrentItems = 0;
        Address = "Не указан";
        ManagerName = "Не назначен";
        IsOpen = true;
    }

    public virtual void GetStorageStatus()
    {
        Console.WriteLine($"Склад: {WarehouseName}");
        Console.WriteLine($"Занято: {CurrentItems} из {StorageCapacity}");
        Console.WriteLine($"Свободно: {StorageCapacity - CurrentItems}");
    }

    public virtual void AddItem(Item item)
    {
        if (CurrentItems < StorageCapacity)
        {
            CurrentItems++;
            Console.WriteLine($"Товар {item.Name} добавлен на склад.");
        }
        else
        {
            Console.WriteLine("На складе нет свободного места.");
        }
    }

    public void AddItem(Item item, int quantity)
    {
        for (int i = 0; i < quantity; i++)
        {
            AddItem(item);
        }
    }

    public virtual void RemoveItem(Item item)
    {
        if (CurrentItems > 0)
        {
            CurrentItems--;
            Console.WriteLine($"Товар {item.Name} удален со склада.");
        }
        else
        {
            Console.WriteLine("На складе нет товаров.");
        }
    }

    public void OpenWarehouse()
    {
        IsOpen = true;
        Console.WriteLine("Склад открыт.");
    }

    public void CloseWarehouse()
    {
        IsOpen = false;
        Console.WriteLine("Склад закрыт.");
    }

    public void ShowManager()
    {
        Console.WriteLine($"Ответственный: {ManagerName}");
    }
}

class PersonalInventory : Inventory
{
    public string OwnerName { get; set; }
    public string OwnerPhone { get; set; }
    public bool HasInsurance { get; set; }

    public PersonalInventory(int id, string name, int capacity, string owner)
        : base(id, name, capacity)
    {
        OwnerName = owner;
        OwnerPhone = "Не указан";
        HasInsurance = false;
    }

    public override void GetStorageStatus()
    {
        Console.WriteLine($"Личный склад владельца: {OwnerName}");
        base.GetStorageStatus();
    }

    public void AddInsurance()
    {
        HasInsurance = true;
        Console.WriteLine("Страховка оформлена.");
    }

    public void ShowOwner()
    {
        Console.WriteLine($"Владелец: {OwnerName}");
    }

    public void ChangeOwner(string name)
    {
        OwnerName = name;
    }
}

class GroupInventory : Inventory
{
    public string ProductGroup { get; set; }
    public int GroupCount { get; set; }
    public string ResponsiblePerson { get; set; }

    public GroupInventory(int id, string name, int capacity, string group)
        : base(id, name, capacity)
    {
        ProductGroup = group;
        GroupCount = 0;
        ResponsiblePerson = "Не назначен";
    }

    public override void AddItem(Item item)
    {
        if (CurrentItems < StorageCapacity)
        {
            CurrentItems++;
            GroupCount++;
            Console.WriteLine($"Товар {item.Name} добавлен в группу {ProductGroup}.");
        }
        else
        {
            Console.WriteLine("На складе нет свободного места.");
        }
    }

    public void ShowGroup()
    {
        Console.WriteLine($"Группа товаров: {ProductGroup}");
    }

    public void ChangeGroup(string group)
    {
        ProductGroup = group;
    }

    public void ShowResponsiblePerson()
    {
        Console.WriteLine($"Ответственный: {ResponsiblePerson}");
    }
}

class AutomatedInventory : Inventory
{
    public int AutomationLevel { get; set; }
    public int RobotCount { get; set; }
    public int SensorCount { get; set; }

    public AutomatedInventory(int id, string name, int capacity, int level)
        : base(id, name, capacity)
    {
        AutomationLevel = level;
        RobotCount = 2;
        SensorCount = 5;
    }

    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine($"Уровень автоматизации: {AutomationLevel}");
    }

    public void AddRobot()
    {
        RobotCount++;
    }

    public void AddSensor()
    {
        SensorCount++;
    }

    public void ShowAutomationInfo()
    {
        Console.WriteLine($"Роботы: {RobotCount}, датчики: {SensorCount}");
    }

    public void ChangeAutomationLevel(int level)
    {
        AutomationLevel = level;
    }
}

// Обобщенный класс
class Storage<T>
{
    public T Item { get; set; }
    public int Quantity { get; set; }

    public Storage(T item, int quantity)
    {
        Item = item;
        Quantity = quantity;
    }

    public void ShowInfo()
    {
        Console.WriteLine($"Содержимое: {Item}, количество: {Quantity}");
    }

    public void ChangeQuantity(int quantity)
    {
        Quantity = quantity;
    }
}

Item phone = new Item("Телефон", 50000, "Samsung", 200);
Item laptop = new Item("Ноутбук", 75000, "Lenovo", 1500);

PersonalInventory personal =
    new PersonalInventory(1, "Личный склад", 10, "Артур");

GroupInventory group =
    new GroupInventory(2, "Групповой склад", 20, "Электроника");

AutomatedInventory automated =
    new AutomatedInventory(3, "Автоматизированный склад", 50, 3);

personal.AddItem(phone);
group.AddItem(laptop, 2);
automated.AddItem(phone);

Console.WriteLine();

Inventory inventory = group;
inventory.GetStorageStatus();

Console.WriteLine();

personal.GetStorageStatus();

Console.WriteLine();

automated.RemoveItem(phone);

Console.WriteLine();

phone.ShowInfo();
phone.ShowInfo(true);

Console.WriteLine($"Стоимость трех телефонов: {phone.GetTotalPrice(3)} руб.");

Console.WriteLine();

Storage<Item> itemStorage = new Storage<Item>(phone, 5);
itemStorage.ShowInfo();

Storage<string> nameStorage = new Storage<string>("Телефоны", 10);
nameStorage.ShowInfo();
```
