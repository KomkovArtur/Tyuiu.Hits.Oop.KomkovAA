```

using System;
using System.Collections.Generic;
using System.Linq;

abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, цена: {Price} руб.");
    }

    public override string ToString()
    {
        return $"{Name} — {Price} руб.";
    }

    public static bool operator ==(Component a, Component b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(Component a, Component b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is Component component && this == component;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

interface IInstallable
{
    void Install();
}

class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string name, decimal price, int cores)
        : base(name, price)
    {
        Cores = cores;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Процессор: {Name}, ядер: {Cores}, цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Процессор {Name} установлен.");
    }
}

class RAM : Component, IInstallable
{
    public int Capacity { get; set; }

    public RAM(string name, decimal price, int capacity)
        : base(name, price)
    {
        Capacity = capacity;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Оперативная память: {Name}, объем: {Capacity} ГБ, цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Оперативная память {Name} установлена.");
    }
}

class GPU : Component, IInstallable
{
    public int Memory { get; set; }

    public GPU(string name, decimal price, int memory)
        : base(name, price)
    {
        Memory = memory;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Видеокарта: {Name}, память: {Memory} ГБ, цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта {Name} установлена.");
    }
}

class HDD : Component, IInstallable
{
    public int Capacity { get; set; }

    public HDD(string name, decimal price, int capacity)
        : base(name, price)
    {
        Capacity = capacity;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Жесткий диск: {Name}, объем: {Capacity} ГБ, цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"Жесткий диск {Name} установлен.");
    }
}

class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component)
        where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент добавлен: {component}");
    }

    public void RemoveComponent<T>(T component)
        where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine($"Компонент удален: {component.Name}");
        }
        else
        {
            Console.WriteLine("Такого компонента нет в компьютере.");
        }
    }

    public decimal GetTotalPrice()
    {
        return components.Sum(component => component.Price);
    }

    public void ShowComponents()
    {
        Console.WriteLine("\nКомпоненты компьютера:");

        foreach (Component component in components)
        {
            component.GetInfo();
        }
    }
}

static class ComputerFactory
{
    public static Computer CreateComputer<T1, T2>(T1 first, T2 second)
        where T1 : Component, IInstallable
        where T2 : Component, IInstallable
    {
        Computer computer = new Computer();

        computer.AddComponent(first);
        computer.AddComponent(second);

        return computer;
    }
}

class Program
{
    static void Main()
    {
        CPU cpu = new CPU("Intel Core i5", 20000, 6);
        RAM ram = new RAM("Kingston", 5000, 16);
        GPU gpu = new GPU("NVIDIA RTX 4060", 35000, 8);
        HDD hdd = new HDD("Seagate", 4000, 1000);

        Computer computer = new Computer();

        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(gpu);
        computer.AddComponent(hdd);

        computer.ShowComponents();

        Console.WriteLine($"\nОбщая стоимость: {computer.GetTotalPrice()} руб.");

        Console.WriteLine("\nСравнение компонентов:");
        Console.WriteLine(cpu == ram);
        Console.WriteLine(cpu != gpu);

        Console.WriteLine("\nУдаление компонента:");
        computer.RemoveComponent(hdd);

        Console.WriteLine($"\nСтоимость после удаления: {computer.GetTotalPrice()} руб.");

        Console.WriteLine("\nСоздание компьютера через фабрику:");
        Computer officeComputer = ComputerFactory.CreateComputer(
            new CPU("Intel Core i3", 12000, 4),
            new RAM("Crucial", 3000, 8)
        );

        officeComputer.ShowComponents();
        Console.WriteLine($"Стоимость офисного компьютера: {officeComputer.GetTotalPrice()} руб.");
    }
}
```
