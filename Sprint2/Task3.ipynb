```
using System;

class Item
{
    public string Name { get; set; }
    public string Manufacturer { get; set; }
    public decimal Price { get; set; }
    public int Weight { get; set; }

    public Item(string name, string manufacturer, decimal price, int weight)
    {
        Name = name;
        Manufacturer = manufacturer;
        Price = price;
        Weight = weight;
    }

    public void ShowItemInfo()
    {
        Console.WriteLine($"Товар: {Name}");
        Console.WriteLine($"Производитель: {Manufacturer}");
        Console.WriteLine($"Цена: {Price} руб.");
        Console.WriteLine($"Вес: {Weight} г.");
    }

    public void ChangePrice(decimal newPrice)
    {
        Price = newPrice;
    }

    public void ChangeWeight(int newWeight)
    {
        Weight = newWeight;
    }

    public decimal GetPrice()
    {
        return Price;
    }
}

class Inventory
{
    public int WarehouseId { get; set; }
    public string WarehouseName { get; set; }
    public int StorageCapacity { get; set; }
    protected int CurrentItems { get; set; }

    public string Address { get; set; }
    public string ManagerName { get; set; }
    public bool IsOpen { get; set; }
    public int SecurityLevel { get; set; }

    public Inventory(
        int id,
        string name,
        int capacity,
        string address,
        string managerName)
    {
        WarehouseId = id;
        WarehouseName = name;
        StorageCapacity = capacity;
        CurrentItems = 0;
        Address = address;
        ManagerName = managerName;
        IsOpen = true;
        SecurityLevel = 1;
    }

    public virtual void GetStorageStatus()
    {
        Console.WriteLine($"Склад: {WarehouseName}");
        Console.WriteLine($"Занято: {CurrentItems} из {StorageCapacity}");
        Console.WriteLine($"Свободно: {StorageCapacity - CurrentItems}");
    }

    public virtual void AddItem(Item item)
    {
        if (CurrentItems < StorageCapacity)
        {
            CurrentItems++;
            Console.WriteLine($"Товар {item.Name} добавлен на склад {WarehouseName}.");
        }
        else
        {
            Console.WriteLine("На складе нет свободного места.");
        }
    }

    public virtual void RemoveItem(Item item)
    {
        if (CurrentItems > 0)
        {
            CurrentItems--;
            Console.WriteLine($"Товар {item.Name} удален со склада {WarehouseName}.");
        }
        else
        {
            Console.WriteLine("На складе нет товаров.");
        }
    }

    public void TransferItem(Item item, Inventory otherInventory)
    {
        if (CurrentItems > 0)
        {
            if (otherInventory.CurrentItems < otherInventory.StorageCapacity)
            {
                CurrentItems--;
                otherInventory.CurrentItems++;

                Console.WriteLine(
                    $"Товар {item.Name} перемещен со склада {WarehouseName} " +
                    $"на склад {otherInventory.WarehouseName}.");
            }
            else
            {
                Console.WriteLine("На другом складе нет свободного места.");
            }
        }
        else
        {
            Console.WriteLine($"На складе {WarehouseName} нет товаров.");
        }
    }

    public void OpenWarehouse()
    {
        IsOpen = true;
        Console.WriteLine($"Склад {WarehouseName} открыт.");
    }

    public void CloseWarehouse()
    {
        IsOpen = false;
        Console.WriteLine($"Склад {WarehouseName} закрыт.");
    }

    public void ChangeSecurityLevel(int level)
    {
        SecurityLevel = level;
        Console.WriteLine($"Уровень безопасности: {SecurityLevel}");
    }

    public void ShowManager()
    {
        Console.WriteLine($"Ответственный: {ManagerName}");
    }
}

class PersonalInventory : Inventory
{
    public string OwnerName { get; set; }
    public string OwnerPhone { get; set; }
    public bool HasInsurance { get; set; }
    public int RoomNumber { get; set; }

    public PersonalInventory(
        int id,
        string name,
        int capacity,
        string address,
        string managerName,
        string owner,
        string phone,
        int room)
        : base(id, name, capacity, address, managerName)
    {
        OwnerName = owner;
        OwnerPhone = phone;
        HasInsurance = false;
        RoomNumber = room;
    }

    public override void GetStorageStatus()
    {
        Console.WriteLine($"Владелец склада: {OwnerName}");
        Console.WriteLine($"Склад: {WarehouseName}");
        Console.WriteLine($"Занято: {CurrentItems} из {StorageCapacity}");
        Console.WriteLine($"Свободно: {StorageCapacity - CurrentItems}");
    }

    public void AddInsurance()
    {
        HasInsurance = true;
        Console.WriteLine("Страховка оформлена.");
    }

    public void ShowOwnerPhone()
    {
        Console.WriteLine($"Телефон владельца: {OwnerPhone}");
    }

    public void ShowRoom()
    {
        Console.WriteLine($"Номер помещения: {RoomNumber}");
    }

    public void ChangeOwner(string newOwner)
    {
        OwnerName = newOwner;
        Console.WriteLine($"Новый владелец: {OwnerName}");
    }
}

class GroupInventory : Inventory
{
    public string ProductGroup { get; set; }
    public int GroupCount { get; set; }
    public string ResponsiblePerson { get; set; }
    public bool HasSeparateSection { get; set; }

    public GroupInventory(
        int id,
        string name,
        int capacity,
        string address,
        string managerName,
        string group)
        : base(id, name, capacity, address, managerName)
    {
        ProductGroup = group;
        GroupCount = 0;
        ResponsiblePerson = managerName;
        HasSeparateSection = true;
    }

    public override void AddItem(Item item)
    {
        if (CurrentItems < StorageCapacity)
        {
            CurrentItems++;
            GroupCount++;

            Console.WriteLine($"Товар {item.Name} добавлен.");
            Console.WriteLine($"Группа товаров: {ProductGroup}");
        }
        else
        {
            Console.WriteLine("На складе нет свободного места.");
        }
    }

    public void ShowProductGroup()
    {
        Console.WriteLine($"Группа товаров: {ProductGroup}");
    }

    public void ShowResponsiblePerson()
    {
        Console.WriteLine($"Ответственный: {ResponsiblePerson}");
    }

    public void ShowGroupCount()
    {
        Console.WriteLine($"Количество товаров группы: {GroupCount}");
    }

    public void ChangeProductGroup(string newGroup)
    {
        ProductGroup = newGroup;
        Console.WriteLine($"Новая группа: {ProductGroup}");
    }
}

class AutomatedInventory : Inventory
{
    public int AutomationLevel { get; set; }
    public int RobotCount { get; set; }
    public int SensorCount { get; set; }
    public bool AutomaticControl { get; set; }

    public AutomatedInventory(
        int id,
        string name,
        int capacity,
        string address,
        string managerName,
        int level)
        : base(id, name, capacity, address, managerName)
    {
        AutomationLevel = level;
        RobotCount = 2;
        SensorCount = 10;
        AutomaticControl = true;
    }

    public override void RemoveItem(Item item)
    {
        if (CurrentItems > 0)
        {
            CurrentItems--;

            Console.WriteLine($"Товар {item.Name} удален со склада.");
            Console.WriteLine($"Уровень автоматизации: {AutomationLevel}");
        }
        else
        {
            Console.WriteLine("На складе нет товаров.");
        }
    }

    public void AddRobot()
    {
        RobotCount++;
        Console.WriteLine($"Количество роботов: {RobotCount}");
    }

    public void AddSensor()
    {
        SensorCount++;
        Console.WriteLine($"Количество датчиков: {SensorCount}");
    }

    public void StartAutomaticControl()
    {
        AutomaticControl = true;
        Console.WriteLine("Автоматическое управление включено.");
    }

    public void StopAutomaticControl()
    {
        AutomaticControl = false;
        Console.WriteLine("Автоматическое управление выключено.");
    }
}

interface IControllable
{
    void ControlSystem();
}

interface IReportable
{
    void MakeReport();
}

class SmartInventory : AutomatedInventory, IControllable, IReportable
{
    public bool ArtificialIntelligence { get; set; }
    public int CameraCount { get; set; }
    public double Temperature { get; set; }
    public bool RemoteControl { get; set; }

    public SmartInventory(
        int id,
        string name,
        int capacity,
        string address,
        string managerName,
        int level)
        : base(id, name, capacity, address, managerName, level)
    {
        ArtificialIntelligence = true;
        CameraCount = 5;
        Temperature = 20;
        RemoteControl = true;
    }

    public void ControlSystem()
    {
        Console.WriteLine("Система умного склада контролируется автоматически.");
    }

    public void MakeReport()
    {
        Console.WriteLine($"Отчет по складу: {WarehouseName}");
        Console.WriteLine($"Товаров: {CurrentItems}");
        Console.WriteLine($"Уровень автоматизации: {AutomationLevel}");
    }

    public void ChangeTemperature(double newTemperature)
    {
        Temperature = newTemperature;
        Console.WriteLine($"Температура склада: {Temperature} градусов.");
    }

    public void AddCamera()
    {
        CameraCount++;
        Console.WriteLine($"Количество камер: {CameraCount}");
    }
}

Item phone = new Item(
    "Телефон",
    "Samsung",
    50000,
    200
);

Item laptop = new Item(
    "Ноутбук",
    "Lenovo",
    75000,
    1500
);

Item mouse = new Item(
    "Мышь",
    "Logitech",
    3000,
    100
);

PersonalInventory personal = new PersonalInventory(
    1,
    "Личный склад",
    10,
    "ул. Ленина, 10",
    "Иван",
    "Артур",
    "87001234567",
    101
);

GroupInventory group = new GroupInventory(
    2,
    "Групповой склад",
    20,
    "ул. Гагарина, 20",
    "Петр",
    "Электроника"
);

AutomatedInventory automated = new AutomatedInventory(
    3,
    "Автоматизированный склад",
    50,
    "ул. Мира, 30",
    "Алексей",
    3
);

SmartInventory smart = new SmartInventory(
    4,
    "Умный склад",
    100,
    "ул. Центральная, 40",
    "Дмитрий",
    5
);

personal.AddItem(phone);
group.AddItem(laptop);
automated.AddItem(mouse);
smart.AddItem(phone);

Console.WriteLine();

personal.GetStorageStatus();
personal.ShowOwnerPhone();
personal.AddInsurance();

Console.WriteLine();

group.GetStorageStatus();
group.ShowProductGroup();
group.ShowResponsiblePerson();

Console.WriteLine();

automated.GetStorageStatus();
automated.AddRobot();
automated.AddSensor();
automated.StartAutomaticControl();

Console.WriteLine();

smart.GetStorageStatus();
smart.ControlSystem();
smart.MakeReport();
smart.ChangeTemperature(18);
smart.AddCamera();

Console.WriteLine();

personal.TransferItem(phone, group);

Console.WriteLine();

personal.GetStorageStatus();

Console.WriteLine();

group.GetStorageStatus();
```
